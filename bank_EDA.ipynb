```python
# Question:1 Create income bands for customers based on their estimated income, categorizing them into Low, Medium, and High-income groups.
#pd.cut() is a Pandas function used to divide numerical values into ranges (bins/categories)

bins=[0,100000,300000,float('inf')]
labels=['low', 'Med', 'High']

df['Income Band']= pd.cut(df['estimated_income'], bins=bins, labels=labels, right=False)
```

### Numerical Analysis:



```python
numerical_cols = ['estimated_income', 'superannuation_savings', 'credit_card_balance', 'bank_loans', 
                  'bank_deposits', 'checking_accounts', 'saving_accounts', 'foreign_currency_account', 'business_lending'  ]

# Univariate analysis and visualization

plt.figure(figure=(20,20))
for i, col in enumerate(numerical_cols):
    plt.subplot(3,3,i+1)
    sns.histplot(df[col], kde=True)
    plt.title(col)
    plt.tight_layout()
    plt.xticks(rotation=42)
    # plt.ticklabel_format(style='plain', axis='x')
    plt.xlabel(col, labelpad=20)
plt.show()
```

```python
# First check actual column names
print(df.columns.tolist())
```

```python
numerical_cols = ['estimated_income', 'superannuation_savings', 'credit_card_balance', 'bank_loans',
                  'bank_deposits', 'checking_accounts', 'saving_accounts', 'foreign_currency_account', 'business_lending']

fig, axes = plt.subplots(3, 3, figsize=(18, 15))

for i, col in enumerate(numerical_cols):
    ax = axes[i // 3, i % 3]
    
    sns.histplot(df[col], kde=True, ax=ax)
    ax.set_title(col, fontsize=14)
    ax.set_xlabel(col, labelpad=15)
    ax.tick_params(axis='x', rotation=42)
    plt.xlabel(col, labelpad=20)

plt.subplots_adjust(hspace=0.7, wspace=0.4)
plt.show()
```

### Heatmaps

```python
numerical_cols = ['estimated_income', 'superannuation_savings', 'credit_card_balance', 'bank_loans', 
                  'bank_deposits', 'checking_accounts', 'saving_accounts', 'foreign_currency_account', 'business_lending'  ]

correlation_matrix = df[numerical_cols].corr()
plt.figure(figsize=(6,6))
sns.heatmap(correlation_matrix, annot=True, cmap='crest', fmt='.2f')
plt.title('correlation_matrix')
plt.show()
```

```python
# ## Insights of EDA

# 1. The strongest positive correlation occur among 'Bank Deposits' with 'Checking Accounts', 'Saving Accounts' and 'Foreign Currency Account' 
# indicating that customers who mentain high balances in one account type often hold substantial amount/funds accross other accounts as well.
```
